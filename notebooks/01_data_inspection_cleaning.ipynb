#### **Collision identifier data-quality finding**

The 2025 collision dataset contains 101,525 records. Although `collision_ref_no` has 101,525 distinct values, `collision_index` has only 2,390 distinct values.

Inspection of the original CSV shows that some `collision_index` values are represented in shortened scientific notation. Re-importing the column as a string does not recover the full identifier.

The repeated `collision_index` values should not be treated as proof of duplicate collision records. I will retain the original records and investigate the documented linking fields before merging the collision, vehicle and casualty datasets.

##### Explaining Cleaning decissions
**Missing coordinates** : Excluding the two records from the co-ordinate based maping gives us a clean dataset where no value is missing and while checking also we found that it didn't affect the dataset much.The dataset contains two collision records with missing geographical coordinates. These records were retained in the main dataset because their other attributes, including collision severity and vehicle counts, remain useful for non-geographical analysis.
For mapping and coordinate-based analysis, a separate subset containing valid latitude and longitude values will be used. This prevents missing coordinates from affecting geographical visualisations while preserving the original collision information.

**Duplicate identifiers** : `collision_index` represent codes in which few of them approximatley 2390 are unique, as beacause these are codes and each and every code has a specific meaning of the data which identifies the collision type. More importantly, we established that some collision_index values already appear in shortened scientific notation in the source CSV. Those repeated values do not establish that the underlying records are duplicates.
We also discovered that `collision_ref_no` contains 101,525 unique values, supporting the conclusion that we should not remove records based solely on repeated `collision_index` values.

**Categorical codes** : Special values such as -1 should not automatically be deleted or treated as ordinary numerical measurements.
Their meaning must first be checked against the official STATS19 data dictionary. Depending on the documented meaning and analytical objective, they may be retained as explicit categories, classified as unknown or excluded from specific calculations.
These decisions will be documented to ensure the analysis remains transparent and reproducible.
